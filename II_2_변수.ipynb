#**변수**
##**: 데이터를 저장할 수 있는 공간**
변수가 만들어지면 주기억 장치에 해당 데이터가 저장

*   형식 : 변수명 = 데이터 ex) x=10
*   의미 : 변수 x에 10을 할당하라는 의미 (변수 x로 주기억 장치에 저장된 10을 참조할 수 있다.)
 *  변수를 정의할 때 데이터의 형태에 따라 자료형이 자동으로 지정된다.
 *  변수의 초기값은 반드시 지정해야 한다.



```python
a = 10
print(10)
print(type(a))   # type() 자료형을 반환하는 함수이다.
```



### 1.   변수는 한번에 한개의 데이터만 저장한다.


```python
age = 14
print(age)
age = 15
age = 16
print(age)
```

### 2. 파이썬에서는 하나의 문장에서 다중 할당이 가능하다.

```python
a, b = 100, '안녕하세요?'
print(a, b)
c = d = e = 200
print(c, d, e)
```

### 3. 변수명 작성 규칙

*   첫 번째 문자는 영문자, 한글 또는 밑줄( _ ) 로 시작해야 한다.
*  나머지 문자는 영문자, 한글, 밑줄( _ ) , 숫자이어야 한다.
*  if, for와 같은 예약어(특정한 기능으로 사용되도록 미리 약속된 단어)는 사용할 수 없다.
*  공백과 하이픈(-)은 사용할 수 없다.
*  영문자 대소문자를 구별한다.
*  저장된 데이터를 알 수 있도록 의미 있는 이름을 사용하는 것이 좋다.
*  여러 단어로 구성된 변수명을 사용하는 경우에는 각 단어를 밑줄( _ ) 로 연결하거나이어지는 단어를 대문자로 시작하여 가독성을 높이는 것이 좋다(예:sum_result, my_age, sumResult, myAge).

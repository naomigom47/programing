# **모듈**

**특정 기능과 값을 편리하게 사용하도록 미리 만들어 놓은 명령어의 집합**


*   모듈 가져오기 : import 모듈명
*   모듈 사용하기 : 모듈명.기능 또는 모듈명.변수

```python
import datetime #datetime이라는 모듈 전체를 가져옵니다.
from datetime import datetime, date, time #datetime 모듈 안에서 자주 쓰는 datetime, date, time 클래스만 콕 집어서 직접 가져옵니다.
```

## **datetime 모듈**
**날짜 및 시간을 다루는데 사용되는 표준 모듈**

| 구분 | date | time| datetime |
| --- | --- | --- | -- |
| 담는 정보 | 날짜 (년, 월, 일) | 시간  (시, 분, 초, 마이크로초) | 날짜 + 시간  (년, 월, 일, 시, 분, 초...) |
| 주요 용도 | 생일, 기념일, 결제일 등 | 영업 시작 시간, 알람 시각 등 | 로그인 기록, 결제 일시, 타임스탬프 등 |
| 생성 예시 | date(2026, 9, 2) | time(14, 30, 0) | datetime(2026, 9, 2, 14, 30, 0) |




### **date 클래스 (날짜 전용)**

시간 정보는 필요 없고 오직 '**날짜**'만 필요할 때 사용

특정 날짜 생성: date(년, 월, 일)

```python
# 2026년 12월 25일 지정
dt1 = date(2026, 12, 25)
print(dt1)

# 개별요소 접근
print(dt1.year)
print(dt1.month)
print(dt1.day)
```

### **datetime 클래스 (날짜 + 시간)**

날짜와 시간을 동시에 다뤄야 하는 대부분의 경우에 사용

특정 날짜/시간 생성: datetime(년, 월, 일, 시, 분, 초, 마이크로초)

```python
dt2 = datetime(2026, 9, 3, 7, 30, 0)
print(dt2)

# 개별 요소 접근
print(dt2.year)
print(dt2.month)
print(dt2.day)
print(dt2.hour)
print(dt2.minute)
print(dt2.second)
```

### **time 클래스 (시간 전용)**

날짜 없이 하루 중 특정 '**시각**'만 표현할 때 사용

특정 시간 생성: time(시, 분, 초, 마이크로초)

```python
dt3 = time(7, 30, 0)
print(dt3)

# 개별 요소 접근
print(dt3.hour)    # 7
print(dt3.minute)  # 30
print(dt3.second)  # 0
```

### **현재 날짜 및 시간 가져오기**

```python
now = datetime.now()
print(now)

today = date.today()
print(today)
```

### **클래스 간 변환 방법**

datetime 객체에서 날짜만 추출하거나, date와 time을 합치는 변환 작업은 자주 사용됩니다.

```python
# [변환 1] datetime -> date, time 으로 분리하기
now = datetime.now()

current_date = now.date()  # date 객체 추출
current_time = now.time()  # time 객체 추출

print(current_date)  # 2026-09-02
print(current_time)  # 21:40:12.345678


# [변환 2] date + time -> datetime 으로 합치기
d = date(2026, 9, 2)
t = time(15, 0, 0)

combined = datetime.combine(d, t)
print(combined)
```

실행 결과를 참고하여 특정 날짜를 입력하면 오늘부터 며칠이 남았는지를 계산하여 출력하는 '디데이(D-day) 프로그램'을 작성해 보자.



*   사용자로부터 연, 월, 일을 숫자 형태로 입력받기
*   입력된 날짜와 오늘 날짜 사이의 일수 계산하기
*   출력 형식에 맞게 출력하기



```python
print('<디데이(D-day) 프로그램>')
print( )
year = int(input('연도를 입력하세요: '))
month = int(input('월을 입력하세요: '))
day = int(input('일을 입력하세요: '))
dt4 = date(year, month, day)
today = date.today( )                # today = 오늘 날짜 저장
dday = (dt4 - today).days            # dday = 두 날짜 사이의 일수 계산하여 저장
print(f'{year}년 {month}월 {day}일은 {dday}일 남았습니다.')
```

회전 초밥집에서는 담긴 접시의 색에 따라 초밥의 가격이 다르게 계산된다. 다음을 참고하여 회전 초밥의 비용을 영수증 형식으로 출력하는 아래의 프로그램을 완성해 보자.


image.png

```python
red_price = 1000
yellow_price = 2000
blue_price = 3000
green_price = 4000
red_count = int(input('빨간색 접시의 개수를 입력하세요: '))
yellow_count = int(input('노란색 접시의 개수를 입력하세요: '))
blue_count = int(input('파란색 접시의 개수를 입력하세요: '))
green_count = int(input('초록색 접시의 개수를 입력하세요: '))

red_total = red_price * red_count
yellow_total = yellow_price * yellow_count
blue_total = blue_price * blue_count

green_total = green_price * green_count
total_price = red_total + yellow_total + blue_total + green_total
print( )
print('===== 회전 초밥 영수증 =====')
print(f'접시\t단가\t개수\t합계')
print(f'빨간색\t{red_price}\t{red_count}\t{red_total}')
print(f'노란색\t{yellow_price}\t{yellow_count}\t{yellow_total}')
print(f'파란색\t{blue_price}\t{blue_count}\t{blue_total}')
print(f'초록색\t{green_price}\t{green_count}\t{green_total}')
print(f'총 가격\t\t\t{total_price}')
print('============================')
```

## **print() 함수에서 출력형식 지정하기**

print( ) 함수에서 출력 형식을 지정하는 방법

*   % 연산자 방식
*   format( ) 함수 방식
*   f-string 방식

출력 형식을 활용하면 값이 변경되어도 일관성 있는 출력 가능

출력에 사용되는 값은 print( ) 함수에 바로 입력하거나 변수에 대입하여 저장 가능





### 1. % 연산자 방식 (C언어 스타일)

문자열 내부의 %d 자리에 뒤에 나오는 % 20의 값인 20이 대입됩니다.

서식 문자 종류
*   %d: 정수 (Decimal)
*   %f: 실수 (Float)
*   %s: 문자열 (String)

가장 고전적인 방식이며, 타입에 맞는 서식 문자(%d, %f 등)를 구분해서
써줘야 합니다.


```python
print('현재 온도는 %d도 입니다.'%20)
```

### 2. str.format() 메서드를 이용한 포맷팅

문자열 안의 빈 중괄호 {} 위치에 .format() 괄호 안에 전달된 값(20)이 순서대로 채워집니다.

파이썬 3.0부터 도입되었으며, %d나 %s처럼 데이터 타입을 일일이 지정하지 않고 중괄호 {}만 쓰면 알아서 변환되므로 훨씬 편리합니다.



```python
print('현재 온도는 {}도 입니다.'.format(20))
```


### 3. f-string (f-문자열) 포맷팅

문자열 시작인 따옴표 앞에 f 접두사를 붙이고, 중괄호 {} 안에 넣고 싶은 값이나 변수(또는 수식)를 직접 작성합니다.

파이썬 3.6부터 도입된 최신 방식으로, 가독성이 매우 높고 처리 속도도 가장 빠르기 때문에 현재 파이썬에서 가장 권장되는 직관적인 출력 방식입니다.



```python
print(f'현재 온도는 {20}도 입니다.')
```

### 값을 2개 사용한 예시

```python
a = 20
b = 50
print('현재 온도는 %d도, 습도는 %d%% 입니다.'%(a, b))
print('현재 온도는 {}도, 습도는 {}% 입니다.'.format(a, b))
print(f'현재 온도는 {a}도, 습도는 {b}% 입니다.')
```

### 소수점 표현 예시

**1. % 서식 문자를 이용한 소수점 제어**

%d: 정수 c (90)가 들어갑니다.

%0.2f: 실수 d를 소수점 둘째 자리까지 표시하라는 의미입니다.

.2는 소수점 아래 2자리까지 출력을 뜻하며, 셋째 자리에서 자동 반올림되어 3.14가 됩니다.


```python
c = 90
d = 3.141592
print('직각은 %d도, 원주율은 %0.2f 입니다.'%(c, d))
```

**2. str.format() 메서드와 인덱스/자릿수 지정**

{0}: .format(c, d)에서 0번째 인자(변수 c)를 가리킵니다.

{1:0.2f}: 1번째 인자(변수 d)를 가져와 콜론(:) 뒤의 서식 지정자 0.2f를 적용해 소수점 둘째 자리까지 출력합니다.




```python
print('직각은 {0}도, 원주율은 {1:0.2f} 입니다.'.format(c, d))
```

**3. f-string을 이용한 소수점 제어**

{d:0.2f}: 변수명 d 뒤에 콜론(:)을 붙이고 바로 0.2f 표현식을 써서 소수점 둘째 자리까지 제한합니다.

0.2f의 의미: f는 실수(Float)를 뜻하며, .2는 소수점 둘째 자리까지 출력(셋째 자리에서 반올림)하라는 규칙입니다.

콜론(:)의 역할: .format()이나 f-string에서는 중괄호 안에서 **변수:서식** 형태로 구분할 때 콜론(:)을 사용합니다.




```python
print(f'직각은 {c}도, 원주율은 {d:0.2f}입니다.')
```

### **print( ) 함수에서 이스케이프 문자 사용하기**

print( ) 함수에서 내용 일부를 줄 바꿈을 하여 출력하거나 특수 문자를 표현하고자 하는 경우에는 이스케이프 문자를 사용할 수 있다.



```python
print('동해 물과 \'백두산\'이')
print('마르고 닳도록')
print('하느님이 보우하사 \n우리나라 만세')
print('무궁화 삼천리', end = ' ')
print('화려 강산')
print('대한 사람, "대한"으로 길이 보전하세.')
print('2026', '05', '20', sep='-')
```

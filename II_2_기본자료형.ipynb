## **숫자형**
*  숫자형 값은 덧셈, 뺄셈, 곱셈, 나눗셈의 산술 연산 가능
*  정수와 실수를 연산한 경우 결과는 실수 형태가 됨.
*  파이썬 3부터 나눗셈 연산자(/)의 작동 방식이 "결과의 정확성"을 유지하도록 설계되어 **float**으로 일관되게 반환
*  소수점을 떼고 정수 형태의 결과만 얻고 싶다면 **//** (소수점 버림 나눗셈 연산자)를 사용

```python
a = 6
b = 1.5
c = 1 + 2j
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(a + 1, type(a + 1)) # 정수 + 정수 = 정수
print(a / b, type(a / b)) # 정수 / 실수 = 실수
print(a / 2, type(a / 2)) # 정수 / 정수 = 실수
print(a // 2, type(a // 2)) # 정수 // 정수 = 정수
```

## **불형**
*  참과 거짓의 값을 저장하는 자료형
*  참을 의미하는 **True**와 거짓을 의미하는 **False** 두가지 갑을 가짐.
*  True와 False는 예약어로 **첫 문자를 항상 대문자**로 작성해야함.
*  논리연산과 비교 연산의 결과를 저장하는데 사용

```python
a = True
b = False
print(a, type(a))
print(b, type(b))
print(10 < 20, type(10 < 20))
print(20 < 10, type(20 < 10))
```

변수에 저장된 값이 참인지 거짓인지 판단하여 출력하는 데 사용

```python
a = 'Hello'
b = ''
c = 10
d = 0
print(bool(a)) # 비어있지 않은 문자열. 내용이 존재하므로 True로 평
print(bool(b)) # 따옴표 안에 아무것도 없는 빈 문자열(empty string). 길이(length)가 0인 빈 데이터이므로 False로 평가
print(bool(c)) # 숫자형에서는 0이 아닌 모든 숫자(양수, 음수, 소수 등)를 True로 평가
print(bool(d)) # 0(또는 0.0, 0j 등)은 완벽한 '없음/무'의 상태를 의미하므로 False로 평가
```

# **문자열(string)형**
*  문자들의 나열을 의미하는 것으로 값을 **작은따옴표**(') 또는 **큰따옴표**(")로 감싸면 문자열형으로 저장
*  작은따옴표와 큰따옴표의 짝을 맞추지 않고 **혼용해서 사용할 수 없음**
*  문자열 안에 따옴표를 포함시키고자 할 때는 바깥쪽과 다른 종류의 따옴표를 사용하거나 바깥쪽에 **따옴표 3개를 연속으로 사용**(''')해야 함.
.

```python
a = '100'
b = 'Python'
c = 'name is "짱구"' # c = 'name is '짱구''(오류 발생)
d = '''name is "코난"'''
print(a, type(a))
print(b, type(b))
print(c, type(c))
print(d, type(d))
```

## **문자열 연산**
* 문자열 : 산술 연산 중에 덧셈과 곱셈 연산자 사용 가능
* 문자열의 덧셈 : 문자열과 문자열을 연결
* 문자열의 곱셈 : 정수와 함께 사용하여 문자열을 특정 횟수만큼 반복
* 문자에 비교 연산자를 사용하면 문자의 크기를 철자 순으로 비교한 결과를 불형으로 알려 줌.
* in과 not in을 통해 문자열 변수에 특정 문자열이 포함된 여부도 불형으로 알 수 있음.


```python
a = 'hello '
b = 'world'
print(a + b)
print(a * 3)
print(a < b)   # 두 문자열을 사전순(알파벳순 / 아스키코드값)으로 비교
print('h' in a)
print('wor' not in b)
```

## **문자열 인덱싱과 슬라이싱**
* 문자열: 포함된 각 문자마다 인덱스 번호가 부여되어 있음.
* 특정한 위치의 단일 문자에 접근하는 인덱싱과 특정 범위에 있는 부분 문자열에 접근하는 슬라이싱 가능
* 인덱스 번호는 0부터 시작하고 공백 문자에도 부여되며, 맨 끝의 번호는 –1로도 표현할 수 있음.
* 변수 a에 ‘Hello Python’을 저장하면 인덱스가 다음과 같이 부여됨.
image.png


```python
# 한 문자만 인덱싱 : 변수명[인덱스번호]
a = 'Hello Python'
print(a[0])
print(a[6])
print(a[-1])
```

```python
# 연속된 부분의 문자열을 슬라이싱 : 변수명[시작인덱스:종료인덱스]
a = 'Hello Python'
print(a[0:5])             # 0 ~ 4
print(a[6:])              # 6 ~ 끝
print(a[:5])              # 0 ~ 4
print(a[6:-4])            # 6 ~ 끝에서 5번째
print(a[0:9:2])           # 0 ~ 8(2간격)
```
